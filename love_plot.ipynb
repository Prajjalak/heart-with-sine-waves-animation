```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
```

```python
# Define parameters
N = 100  # Maximum value of n
frames = list(range(N + 1)) + ([N] * 25) + list(range(N - 1, -1, -1))  # n sweeps 0 -> N (hold) -> 0
x = np.arange(-1., 1.001, 0.001)  # x-range: [-1, 1]
```

```python
# Create figure
fig = plt.figure(figsize=(4, 5))
gs = fig.add_gridspec(2, 1, height_ratios=[1, 4], hspace=0)

title = fig.add_subplot(gs[0])
title.axis("off")
text = title.text(
    0.5, 0.5,  # x, y
    "$y = x^{\\frac{2}{3}} + \\sqrt{1-x^2}\\sin{(n \\pi x)}$" + f"\nn = {N}",
    fontsize=16, ha="center", va="center"
)

ax = fig.add_subplot(gs[1])
ax.xaxis.set_ticks_position("both")
ax.yaxis.set_ticks_position("both")
ax.tick_params(direction="in", length=5, width=1)
ax.xaxis.set_tick_params(which="both", labeltop=True)  # Top tick labels
ax.yaxis.set_tick_params(which="both", labelright=True)  # Right tick labels
ax.spines["top"].set_visible(True)
ax.spines["right"].set_visible(True)

ax.set_xlim(-1.2, 1.2)
ax.set_ylim(-1.2, 1.7)
ax.grid(True)

line, = ax.plot([], [], "xkcd:rose red", lw=2)

plt.tight_layout()
```

```python
# Function definition: y = f(x)
def y(x, n):
    return np.cbrt(x**2) + (np.sqrt(1 - x**2)*np.sin(n*np.pi*x))
```

```python
# Plotting function
def plot(n):
    line.set_data(x, y(x, n))
    text.set_text("$y = x^{\\frac{2}{3}} + \\sqrt{1-x^2}\\sin{(n \\pi x)}$" + f"\nn = {n}\n\n")
    return line, text
```

```python
# Create animation
ani = animation.FuncAnimation(fig, plot, frames=frames, blit=True)
```

```python
# Save as GIF
ani.save("love.gif", writer="pillow", fps=20, dpi=100)
```
